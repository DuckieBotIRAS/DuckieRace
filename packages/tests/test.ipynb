```python
import numpy as np

def calc(dx,dy):
    error = np.sqrt(dx**2 + dy**2)
    scale = 1 / error
    if dx >= 0 and dy >= 0:
        return np.asin(dy*scale) / (np.pi *2) * 360 
    elif dx >= 0 and dy < 0:
        return 360 + np.asin(dy*scale) / (np.pi *2) * 360 
    elif dx < 0 and dy >= 0:
        return 180 - np.asin(dy*scale) / (np.pi *2) * 360 
    else:
        return 180 - np.asin(dy*scale) / (np.pi *2) * 360 
```

```python
def getch():
    import sys, termios

    fd = sys.stdin.fileno()
    orig = termios.tcgetattr(fd)

    new = termios.tcgetattr(fd)
    new[3] = new[3] & ~termios.ICANON
    new[6][termios.VMIN] = 1
    new[6][termios.VTIME] = 0

    try:
        termios.tcsetattr(fd, termios.TCSAFLUSH, new)
        return sys.stdin.read(1)
    finally:
        termios.tcsetattr(fd, termios.TCSAFLUSH, orig)
```
