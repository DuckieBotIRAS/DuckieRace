```python
from turtle import *

pensize(4)  # Make the lines thicker than normal.

forward(200)  # Move the turtle forward 200 steps.
left(90)  # Turn the turtle left by 90 degrees.

# Move forward and turn three more times:
forward(200)
left(90)
forward(200)
left(90)
forward(200)
left(90)

done() 
```

```python
def draw_multi(num_edges):
    degrees_total = 180 * (num_edges - 2)
    alpha = 180 - (degrees_total / num_edges)
    pensize(4) 

    for i in range(num_edges):
        forward(100)
        left(alpha)

    done()

draw_multi(7)
```

```python
pensize(4) 
for i in range(50):
    forward(i * 5)
    left(30)

done()
```

```python
speed('fastest')

setheading(0)  # Face right.
for i in range(20):
    circle(i * 10)

setheading(180)  # Face right.
for i in range(20):
    circle(i * 10)

done()
```

```python
diagonal = 2**(1/2) * 100

speed('slow')

for i in range(3):
    forward(100)
    left(90)
    forward(100)
    left(135)
    forward(diagonal)
    right(135)
    forward(100)
    right(45)
    forward(diagonal / 2)
    right(90)
    forward(diagonal / 2)
    right(135)
    forward(100)
    left(135)
    forward(diagonal)
    left(45)

done()
```
